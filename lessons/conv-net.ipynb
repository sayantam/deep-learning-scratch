```python
import numpy as np

from dls.cnn import Conv2D, ReLU
from dls.dataset import mnist_train_test_split
from dls.dnn import NeuralNetwork, Dense, Linear, Trainer
from dls.dnn_ext import Tanh, SoftmaxCrossEntropyLoss, SGDMomentum
from dls.model_error import calc_accuracy_model
```

```python
mtts = mnist_train_test_split()
X_train, y_train, X_test, y_test = mtts.X_train, mtts.y_train, mtts.X_test, mtts.y_test
```

```python
%load_ext autoreload
%autoreload 2
```

```python
train_mean = np.mean(X_train)
train_std = np.std(X_train)
X_train = (X_train - train_mean) / train_std
X_test = (X_test - train_mean) / train_std
```

```python
X_train_conv, X_test_conv = (
    X_train.reshape(-1, 1, 28, 28),
    X_test.reshape(-1, 1, 28, 28),
)
```

```python
num_labels = len(y_train)
train_labels = np.zeros((num_labels, 10))
for i in range(num_labels):
    train_labels[i][y_train[i]] = 1

num_labels = len(y_test)
test_labels = np.zeros((num_labels, 10))
for i in range(num_labels):
    test_labels[i][y_test[i]] = 1
```

```python
model = NeuralNetwork(
    layers=[
        Conv2D(
            out_channels=16,
            param_size=5,
            dropout=0.8,
            weight_init="glorot",
            flatten=True,
            activation=Tanh(),
        ),
        Dense(neurons=10, activation=Linear()),
    ],
    loss=SoftmaxCrossEntropyLoss(),
    seed=20190402,
)

trainer = Trainer(model, SGDMomentum(learning_rate=0.1, momentum=0.9))
trainer.fit(
    X_train_conv,
    train_labels,
    X_test_conv,
    test_labels,
    epochs=1,
    eval_every=1,
    seed=20190402,
    batch_size=60,
    conv_testing=True,
)
calc_accuracy_model(model, X_test_conv, y_test, inference=True)
```
